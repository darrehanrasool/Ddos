# DDoS-AI: 03 - Feature Engineering & Shannon Entropy Analysis

Analysis of network statistical dynamics, TCP flag asymmetries, and information-theoretic metrics.

### Objectives:
1. Compute SYN-to-ACK ratios to isolate half-open connection floods.
2. Analyze packet inter-arrival time (IAT) variance across volumetric attacks.
3. Compute Shannon Entropy across sliding observation windows.
4. Measure Pearson feature cross-correlations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ddos_ai.data.feature_engineering import FeatureEngineer

df = pd.read_csv("../data/raw/network_traffic/cic_ddos2019_sample.csv")
fe = FeatureEngineer()
df_rich = fe.compute_flow_features(df)

plt.figure(figsize=(8, 5))
sns.boxplot(data=df_rich, x="Label", y="Packet_Asymmetry_Ratio")
plt.title("Packet Asymmetry Ratio by Attack Category", fontsize=13, fontweight="bold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
