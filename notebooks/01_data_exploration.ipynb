# DDoS-AI: 01 - Exploratory Data Analysis (EDA)

Comprehensive exploratory analysis of network traffic flow datasets (CIC-DDoS2019 / BoT-IoT).

### Objectives:
1. Load raw bidirectional network flow records.
2. Inspect flow durations, packet counts, and byte rates.
3. Visualize class balance between Benign traffic and 12 reflection/amplification attack classes.
4. Profile packet length distributions across TCP and UDP protocol categories.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load sample flow dataset
data_path = Path("../data/raw/network_traffic/cic_ddos2019_sample.csv")
if not data_path.exists():
    from ddos_ai.data.dataset_downloader import DatasetDownloader
    downloader = DatasetDownloader(raw_dir="../data/raw/network_traffic")
    downloader.generate_synthetic_benchmark()

df = pd.read_csv(data_path)
print(f"Dataset Dimensions: {df.shape}")
df.head()
```

```python
# Inspect target distribution
plt.figure(figsize=(10, 5))
df["Label"].value_counts().plot(kind="bar", color="#1f77b4", edgecolor="black")
plt.title("DDoS Attack Type Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Traffic Category")
plt.ylabel("Flow Count")
plt.xticks(rotation=45, ha="right")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
